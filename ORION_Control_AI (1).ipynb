# ORION Control AI

Sistema Inteligente de Monitoramento e Diagnóstico para Missões Espaciais Experimentais

O ORION Control AI é uma solução em Python que simula o monitoramento de uma missão espacial experimental.

O sistema analisa dados de temperatura, energia, comunicação, oxigênio e estabilidade operacional, gera alertas automáticos e utiliza inteligência artificial para produzir diagnósticos e recomendações de ação.

```python
# ORION Control AI
# Sistema Inteligente de Monitoramento e Diagnóstico para Missões Espaciais

!pip install openai -q

import textwrap
from openai import OpenAI
from google.colab import userdata
```

```python
api_key = userdata.get("OPENAI_API_KEY")

if api_key is None:
    raise ValueError("API Key não encontrada. Crie o segredo OPENAI_API_KEY no menu Secrets do Colab.")

client = OpenAI(api_key=api_key)
```

```python
nome_missao = "ORION-7"
nome_equipe = "ORION Team"

cenarios = {
    "1": {
        "nome": "Operação Normal",
        "temperatura": 24,
        "energia": 88,
        "comunicacao": 95,
        "oxigenio": 97,
        "estabilidade": 93
    },
    "2": {
        "nome": "Missão em Atenção",
        "temperatura": 38,
        "energia": 45,
        "comunicacao": 60,
        "oxigenio": 85,
        "estabilidade": 70
    },
    "3": {
        "nome": "Missão Crítica",
        "temperatura": 95,
        "energia": 12,
        "comunicacao": 18,
        "oxigenio": 65,
        "estabilidade": 30
    }
}
```

```python
def classificar_parametros(dados):
    alertas = []
    risco = 0

    if dados["temperatura"] > 80:
        alertas.append("Temperatura crítica: risco de superaquecimento.")
        risco += 2
    elif dados["temperatura"] > 35:
        alertas.append("Temperatura elevada: monitorar controle térmico.")
        risco += 1

    if dados["energia"] < 20:
        alertas.append("Energia crítica: ativar modo de economia.")
        risco += 2
    elif dados["energia"] < 50:
        alertas.append("Energia baixa: reduzir consumo dos sistemas.")
        risco += 1

    if dados["comunicacao"] < 30:
        alertas.append("Comunicação crítica: tentativa de reconexão necessária.")
        risco += 2
    elif dados["comunicacao"] < 70:
        alertas.append("Comunicação instável: monitorar sinal com a base.")
        risco += 1

    if dados["oxigenio"] < 80:
        alertas.append("Oxigênio crítico: acionar protocolo de suporte à vida.")
        risco += 2
    elif dados["oxigenio"] < 90:
        alertas.append("Oxigênio em atenção: acompanhar consumo.")
        risco += 1

    if dados["estabilidade"] < 40:
        alertas.append("Estabilidade crítica: reduzir operações não essenciais.")
        risco += 2
    elif dados["estabilidade"] < 75:
        alertas.append("Estabilidade reduzida: manter nave em observação.")
        risco += 1

    return alertas, risco


def classificar_missao(risco):
    if risco <= 2:
        return "MISSÃO ESTÁVEL"
    elif risco <= 5:
        return "MISSÃO EM ATENÇÃO"
    else:
        return "MISSÃO CRÍTICA"
```

```python
def gerar_prompt_ia(dados, alertas, risco, classificacao):

    alertas_formatados = "\n".join(alertas) if alertas else "Nenhum alerta."

    prompt = f"""
Você é o ORION Control AI.

Analise a missão espacial abaixo.

Missão: {nome_missao}
Cenário: {dados["nome"]}

Temperatura: {dados["temperatura"]}°C
Energia: {dados["energia"]}%
Comunicação: {dados["comunicacao"]}%
Oxigênio: {dados["oxigenio"]}%
Estabilidade: {dados["estabilidade"]}%

Alertas:
{alertas_formatados}

Risco: {risco}/10
Classificação: {classificacao}

Responda em português usando exatamente os tópicos:

DIAGNÓSTICO:
PRINCIPAIS RISCOS:
RECOMENDAÇÕES:
DECISÃO FINAL:

Limite máximo de 150 palavras.
"""
    return prompt
```

```python
def resposta_ia(prompt):
    resposta = client.responses.create(
        model="gpt-4o-mini",
        input=[
            {
                "role": "system",
                "content": "Você é o ORION Control AI, um sistema inteligente de controle de missão espacial. Responda em português, com linguagem técnica, clara e organizada."
            },
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return resposta.output_text
```

```python
def executar_orion_control():
    print("=" * 70)
    print("ORION CONTROL AI")
    print("Sistema Inteligente de Monitoramento Espacial")
    print("=" * 70)

    print("\nEscolha um cenário:")
    print("1 - Operação Normal")
    print("2 - Missão em Atenção")
    print("3 - Missão Crítica")

    escolha = input("\nDigite o número do cenário: ")

    if escolha not in cenarios:
        print("Opção inválida. Executando cenário crítico como teste.")
        escolha = "3"

    dados = cenarios[escolha]
    alertas, risco = classificar_parametros(dados)
    classificacao = classificar_missao(risco)
    prompt = gerar_prompt_ia(dados, alertas, risco, classificacao)

    print("\nConsultando inteligência artificial...")

    try:
        analise_ia = resposta_ia(prompt)
    except Exception as erro:
        analise_ia = f"Erro ao consultar a IA: {erro}"

    print("\n" + "=" * 70)
    print(f"MISSÃO: {nome_missao}")
    print(f"EQUIPE: {nome_equipe}")
    print(f"CENÁRIO: {dados['nome']}")
    print("=" * 70)

    print(f"Temperatura: {dados['temperatura']}°C")
    print(f"Energia: {dados['energia']}%")
    print(f"Comunicação: {dados['comunicacao']}%")
    print(f"Oxigênio: {dados['oxigenio']}%")
    print(f"Estabilidade: {dados['estabilidade']}%")

    print("\nALERTAS AUTOMÁTICOS:")
    if alertas:
        for alerta in alertas:
            print(f"- {alerta}")
    else:
        print("- Nenhum alerta identificado.")

    print(f"\nPontuação de risco: {risco}/10")
    print(f"Classificação: {classificacao}")

    print("\n" + "=" * 70)
    print("ANÁLISE DA INTELIGÊNCIA ARTIFICIAL")
    print("=" * 70)
    print(textwrap.fill(analise_ia, width=95))
```

```python
executar_orion_control()
```
